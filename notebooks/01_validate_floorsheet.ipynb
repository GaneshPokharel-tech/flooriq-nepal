```python
import pandas as pd
from pathlib import Path
```

```python
csv_path = Path("../data/raw/floorsheet_latest.csv")
print("Exists:", csv_path.exists())
print("Path:", csv_path.resolve())
```

```python
df = pd.read_csv(csv_path)
df.head(5)
```

```python
print("Rows:", len(df))
print("Cols:", df.shape[1])
print("Columns:", list(df.columns))
df.dtypes
```

```python
expected = [
    "contractId",
    "stockSymbol",
    "buyerMemberId",
    "sellerMemberId",
    "contractQuantity",
    "contractRate",
    "contractAmount",
    "businessDate",
    "tradeTime",
    "buyerBrokerName",
    "sellerBrokerName",
    "securityName",
]

missing = [c for c in expected if c not in df.columns]
print("Missing expected columns:", missing)

print("Null % (expected cols):")
print((df[expected].isna().mean() * 100).round(3))
```
